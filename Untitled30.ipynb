```python
img = cv2.imread('hq720.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
img_grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
#img_rgb
img_grey
```

```python
plt.figure(figsize=(8,5))
plt.imshow(img)
plt.title('aura')
plt.show()
```

```python
#linhas , colunas
fig, ax = plt.subplots(2, 2, figsize=(7,4))
#como mudou para uma matriz 2 por 2 tem que mudar nos colchetes tambem

ax[0,0].imshow(img)
ax[0,0].set_title('aura bgr')

ax[0,1].imshow(img_rgb)
ax[0,1].set_title('aura rgb')

ax[1,0].imshow(img_hsv)
ax[1,0].set_title('aura hsv')

ax[1,1].imshow(img_grey, cmap='gray')
ax[1,1].set_title('aura grey')

plt.savefig('mapa de imagem')



plt.show()
```
