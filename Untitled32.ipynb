```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
nome_normal = "normal.jpeg"
nome_clara = "clara.jpeg"
nome_escura = "escura.jpeg"

imagem_normal = cv2.imread(nome_normal)
imagem_clara = cv2.imread(nome_clara)
imagem_escura = cv2.imread(nome_escura)

print("Imagem normal:", imagem_normal.shape)
print("Imagem clara:", imagem_clara.shape)
print("Imagem escura:", imagem_escura.shape)
```

```python
cinza_normal = cv2.cvtColor(imagem_normal, cv2.COLOR_BGR2GRAY)
cinza_clara = cv2.cvtColor(imagem_clara, cv2.COLOR_BGR2GRAY)
cinza_escura = cv2.cvtColor(imagem_escura, cv2.COLOR_BGR2GRAY)

print("Conversão para tons de cinza realizada!")
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cinza_normal, cmap="gray")
plt.title("Imagem bem exposta")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cinza_clara, cmap="gray")
plt.title("Imagem clara demais")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cinza_escura, cmap="gray")
plt.title("Imagem escura demais")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(3, 1, 1)
plt.hist(cinza_normal.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Imagem normal")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.subplot(3, 1, 2)
plt.hist(cinza_clara.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Imagem clara demais")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.subplot(3, 1, 3)
plt.hist(cinza_escura.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Imagem escura demais")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.tight_layout()
plt.show()
```

estão mais concentradas para o branco e menos para o escuro

```python
def corrigir_gamma(imagem, gamma):
    """
    Aplica correção de gamma.
    gamma < 1 -> clareia
    gamma > 1 -> escurece
    """

    imagem_normalizada = imagem / 255.0

    resultado = 255 * (imagem_normalizada ** gamma)

    return resultado.astype(np.uint8)


gamma_escura = 0.5

escura_corrigida = corrigir_gamma(cinza_escura, gamma_escura)

print("Correção Gamma aplicada.")
print("Gamma utilizado:", gamma_escura)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cinza_escura, cmap="gray")
plt.title("Antes - Imagem escura")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(escura_corrigida, cmap="gray")
plt.title("Depois - Gamma = 0,5")
plt.axis("off")

plt.tight_layout()
plt.show()
```

usei a correção de gama para puxar os pixels escuros para o tom mais claro e original


```python
gamma_clara = 2.0

clara_corrigida = corrigir_gamma(cinza_clara, gamma_clara)

print("Correção Gamma aplicada.")
print("Gamma utilizado:", gamma_clara)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cinza_clara, cmap="gray")
plt.title("Antes - Imagem clara")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(clara_corrigida, cmap="gray")
plt.title("Depois - Gamma = 2,0")
plt.axis("off")

plt.tight_layout()
plt.show()
```

fiz a mesma coisa nessa so que usando o inverso da imagem escura

```python
def expandir_contraste(imagem):
    """
    Expande o contraste da imagem.

    L = menor intensidade encontrada
    H = maior intensidade encontrada
    """

    L = imagem.min()
    H = imagem.max()

    if H == L:
        return imagem.copy(), L, H

    resultado = (imagem.astype(float) - L) * 255 / (H - L)

    resultado = resultado.astype(np.uint8)

    return resultado, L, H


bem_exposta_corrigida, L, H = expandir_contraste(cinza_normal)

print("Expansão de contraste aplicada.")
print("L =", L)
print("H =", H)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cinza_normal, cmap="gray")
plt.title("Antes - Imagem bem exposta")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(bem_exposta_corrigida, cmap="gray")
plt.title(f"Depois - Expansão (L={L}, H={H})")
plt.axis("off")

plt.tight_layout()
plt.show()
```

aqui usei a expanção de contraste porque é uma forma menos agressiva de correção de distribuição de intensidade

```python
plt.figure(figsize=(14, 8))

plt.subplot(2, 2, 1)
plt.imshow(cinza_escura, cmap="gray")
plt.title("Antes - Imagem escura")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(escura_corrigida, cmap="gray")
plt.title("Depois - Gamma = 0,5")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.hist(cinza_escura.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Antes")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.subplot(2, 2, 4)
plt.hist(escura_corrigida.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Depois")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.suptitle("Imagem Escura - Correção Gamma", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 8))

plt.subplot(2, 2, 1)
plt.imshow(cinza_clara, cmap="gray")
plt.title("Antes - Imagem clara")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(clara_corrigida, cmap="gray")
plt.title("Depois - Gamma = 2,0")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.hist(cinza_clara.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Antes")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.subplot(2, 2, 4)
plt.hist(clara_corrigida.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Depois")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.suptitle("Imagem Clara - Correção Gamma", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 8))

plt.subplot(2, 2, 1)
plt.imshow(cinza_normal, cmap="gray")
plt.title("Antes - Imagem bem exposta")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(bem_exposta_corrigida, cmap="gray")
plt.title(f"Depois - Expansão (L={L}, H={H})")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.hist(cinza_normal.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Antes")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.subplot(2, 2, 4)
plt.hist(bem_exposta_corrigida.ravel(), bins=256, range=(0, 256))
plt.title("Histograma - Depois")
plt.xlabel("Nível de cinza")
plt.ylabel("Quantidade de pixels")

plt.suptitle("Imagem Bem Exposta - Expansão de Contraste", fontsize=16)

plt.tight_layout()
plt.show()
```

### Conclusão

A operação que fez menos diferença visual foi a expansão de contraste aplicada à imagem normal. Isso ocorreu porque essa imagem já apresentava uma exposição relativamente adequada, com detalhes visíveis e uma distribuição de intensidades mais equilibrada. Dessa forma, a expansão de contraste teve um efeito menor do que as correções Gamma aplicadas às imagens escura e clara, que apresentavam problemas mais evidentes de exposição.
